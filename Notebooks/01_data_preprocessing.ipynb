```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("C:\\Users\\Rak\\Desktop\\Projects\\Restaurant Insights\\Dataset.csv")
df.head(10)
```

```python
df.columns = df.columns.str.strip().str.replace(' ', '_')
df.columns
```

```python
df = df.drop_duplicates()
df.shape
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset=['Cuisines'])
```

```python
df['Average_Cost_for_two'] = pd.to_numeric(df['Average_Cost_for_two'], errors='coerce')
df['Aggregate_rating'] = pd.to_numeric(df['Aggregate_rating'], errors='coerce')
df['Votes'] = pd.to_numeric(df['Votes'], errors='coerce')
```

```python
df = df.dropna(subset=['Average_Cost_for_two', 'Aggregate_rating'])
```

```python
df['Cuisines'] = df['Cuisines'].str.strip()

cuisines = df['Cuisines'].str.split(',').explode()
cuisines = cuisines.str.strip()

top_cuisines = cuisines.value_counts().head(10)
top_cuisines
```

```python
city_counts = df['City'].value_counts().head(10)
city_counts
```

```python
df.to_csv("cleaned_restaurant_data.csv", index=False)
```
